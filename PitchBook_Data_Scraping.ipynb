### Code for extracting all the data from the website of each query present in the row, matching the email with the given email, and adding it to the results if a match is found

```python
from selenium.webdriver.common.by import By
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
import time
import random
from urllib.parse import urlparse
from selenium.webdriver.common.action_chains import ActionChains

# Random delay to simulate human-like behavior
def random_delay():
    time.sleep(random.uniform(4, 12))  # Shorter delay to speed things up slightly

# Scroll down the page to simulate human behavior
def simulate_scrolling(driver):
    try:
        action = ActionChains(driver)
        for _ in range(random.randint(3, 6)):
            action.scroll_by_amount(0, random.randint(200, 500)).perform()
            random_delay()
    except Exception as e:
        print(f"Error while scrolling: {e}")

def extract_all_data(driver, url):
    driver.get(url)
    # simulate_scrolling(driver)
    random_delay()  # Wait for the page to load

    data = {}
    try:
        website_elem = driver.find_element(By.XPATH, "//h5[text()='Website']/following-sibling::a")
        data["Website"] = website_elem.get_attribute('href')
        
        description_elem = driver.find_element(By.XPATH, "//h4[contains(text(), 'Description')]/following-sibling::p")
        data["Description"] = description_elem.text

        items = driver.find_elements(By.CLASS_NAME, 'pp-overview-item')
        for item in items:
            label = item.find_element(By.XPATH, ".//li[contains(@class, 'dont-break')]").text
            value = item.find_element(By.XPATH, ".//span[contains(@class, 'pp-overview-item__title')]").text
            data[label] = value
    except Exception as e:
        print(f"Error extracting data: {e}")
    return data

def domain_matches(url, email_domain):
    try:
        domain = urlparse(url).netloc
        return email_domain in domain
    except Exception:
        return False

def process_queries(data):
    results = []
    driver = get_driver()  # Ensure your driver is configured correctly
    try:
        for query in data["Query"].unique():
            query_data = data[data["Query"] == query]
            query_email = query_data["Query Email"].iloc[0]
            email_domain = query_email.split('@')[-1]

            match_found = False
            for _, row in query_data.iterrows():
                url = row["link"]
                extracted_data = extract_all_data(driver, url)

                if "Website" in extracted_data and domain_matches(extracted_data["Website"], email_domain):
                    results.append({"Query": query, **extracted_data})
                    print(f"[INFO] Match found: {query} | Email Domain: {email_domain} | Website: {extracted_data['Website']}")
                    match_found = True
                    break  # Move to the next unique query once a match is found

            if not match_found:
                print(f"[INFO] No match found for query: {query} | Email Domain: {email_domain}")
    finally:
        driver.quit()

    return results

def process_first_20_queries(data):
    results = []
    driver = get_driver()  # Ensure your driver is configured correctly
    try:
        unique_queries = data["Query"].unique()[20:40]  # Get the first 20 unique queries
        for query in unique_queries:
            query_data = data[data["Query"] == query]
            query_email = query_data["Query Email"].iloc[0]
            email_domain = query_email.split('@')[-1]

            match_found = False
            for _, row in query_data.iterrows():
                url = row["link"]
                extracted_data = extract_all_data(driver, url)

                if "Website" in extracted_data and domain_matches(extracted_data["Website"], email_domain):
                    results.append({"Query": query, **extracted_data})
                    print(f"[INFO] Match found: {query} | Email Domain: {email_domain} | Website: {extracted_data['Website']}")
                    match_found = True
                    break  # Move to the next unique query once a match is found

            if not match_found:
                print(f"[INFO] No match found for query: {query} | Email Domain: {email_domain}")
    finally:
        driver.quit()

    return results
import logging

# Configure logging to capture errors in a file and display in the console
logging.basicConfig(
    level=logging.INFO,
    format='%(asctime)s - %(levelname)s - %(message)s',
    handlers=[
        logging.FileHandler("errors.log"),
        logging.StreamHandler()
    ]
)

def process_next_20_queries(data, start=20, end=40):
    results = []
    driver = get_driver()  # Ensure your driver is configured correctly
    try:
        unique_queries = data["Query"].unique()[start:]  # Get the 20-40th unique queries
        for query in unique_queries:
            query_data = data[data["Query"] == query]
            query_email = query_data["Query Email"].iloc[0]
            email_domain = query_email.split('@')[-1]

            match_found = False
            attempts = 0  # Counter for the number of attempts per query

            for _, row in query_data.iterrows():
                if attempts >= 5:  # Limit to 5 attempts per query
                    logging.info(f"[INFO] Max attempts (5) reached for query: {query}")
                    break

                url = row["link"]
                try:
                    extracted_data = extract_all_data(driver, url)
                    attempts += 1  # Increment attempts counter

                    if "Website" in extracted_data and domain_matches(extracted_data["Website"], email_domain):
                        results.append({"Query": query, **extracted_data})
                        logging.info(f"[INFO] Match found: {query} | Email Domain: {email_domain} | Website: {extracted_data['Website']}")
                        match_found = True
                        break  # Move to the next unique query once a match is found
                except Exception as e:
                    logging.error(f"[ERROR] Failed to process URL {url} for query {query}: {e}")
                    attempts += 1  # Still count this as an attempt
                    continue  # Move to the next URL

            if not match_found:
                logging.info(f"[INFO] No match found for query: {query} | Email Domain: {email_domain} | Attempts: {attempts}")
    finally:
        driver.quit()

    return results



```
